```python
import pandas as pd

# Üç dataseti user_id ve month'a göre outer join ile birleştirme
def merge_three_datasets(merged_call_users_plan, merged_message_users_plan, merged_internet_users_plan):
    """
    Üç dataseti birleştirip yeni bir comprehensive dataset oluşturur
    
    Parameters:
    - merged_call_users_plan: Çağrı verileri içeren DataFrame
    - merged_message_users_plan: Mesaj verileri içeren DataFrame
    - merged_internet_users_plan: İnternet verileri içeren DataFrame
    
    Returns:
    - final_dataset: Birleştirilmiş DataFrame
    """
    
    # Ortak sütunlar listesi
    common_columns = ['user_id', 'month', 'first_name', 'last_name', 'age', 'reg_date', 
                      'churn_date', 'city', 'plan', 'usd_monthly_pay', 'minutes_included', 
                      'messages_included', 'mb_per_month_included', 'usd_per_minute', 
                      'usd_per_message', 'usd_per_gb']
    
    # İlk iki dataseti birleştir (call + message)
    print("Adım 1: Call ve Message datasetleri birleştiriliyor...")
    temp_merged = pd.merge(
        merged_call_users_plan,
        merged_message_users_plan,
        on=common_columns,
        how='outer'
    )
    print(f"Ara sonuç: {temp_merged.shape[0]} satır, {temp_merged.shape[1]} sütun")
    
    # Üçüncü dataseti ekle (+ internet)
    print("\nAdım 2: Internet dataseti ekleniyor...")
    final_dataset = pd.merge(
        temp_merged,
        merged_internet_users_plan,
        on=common_columns,
        how='outer'
    )
    print(f"Final sonuç: {final_dataset.shape[0]} satır, {final_dataset.shape[1]} sütun")
    
    # Sütunları yeniden düzenle
    column_order = [
        # Kullanıcı kimlik bilgileri
        'user_id', 'first_name', 'last_name', 'age', 'city',
        'reg_date', 'churn_date', 'month',
        
        # Plan bilgileri
        'plan', 'usd_monthly_pay', 
        'minutes_included', 'messages_included', 'mb_per_month_included',
        'usd_per_minute', 'usd_per_message', 'usd_per_gb',
        
        # Çağrı kullanım verileri
        'duration', 'duration_minutes', 'extra_call_cost',
        
        # Mesaj kullanım verileri
        'message_count', 'extra_message_cost',
        
        # İnternet kullanım verileri
        'mb_used', 'gb_used', 'extra_internet_cost'
    ]
    
    # Sadece mevcut sütunları seç
    available_columns = [col for col in column_order if col in final_dataset.columns]
    remaining_columns = [col for col in final_dataset.columns if col not in available_columns]
    
    final_dataset = final_dataset[available_columns + remaining_columns]
    
    print("\n✅ Birleştirme tamamlandı!")
    print(f"\nFinal Dataset Bilgileri:")
    print(f"- Toplam satır sayısı: {final_dataset.shape[0]}")
    print(f"- Toplam sütun sayısı: {final_dataset.shape[1]}")
    
    # Eksik veri istatistikleri
    print(f"\n📊 Eksik Veri İstatistikleri:")
    missing_stats = final_dataset.isnull().sum()
    missing_stats = missing_stats[missing_stats > 0].sort_values(ascending=False)
    if len(missing_stats) > 0:
        for col, count in missing_stats.items():
            percentage = (count / len(final_dataset)) * 100
            print(f"  {col}: {count} ({percentage:.2f}%)")
    else:
        print("  Hiç eksik veri yok!")
    
    print(f"\n📋 Sütun listesi: {list(final_dataset.columns)}")
    
    return final_dataset


# Kullanım:
final_df = merge_three_datasets(merged_call_users_plan, merged_message_users_plan, merged_internet_users_plan)

final_df['total_cost'] = final_df['usd_monthly_pay'] + final_df['extra_call_cost'] + final_df['extra_message_cost'] + final_df['extra_internet_cost']
final_df.sample(10)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Matplotlib için Türkçe karakter desteği
plt.rcParams['font.sans-serif'] = ['DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

def plot_distribution_comparison(final_df):
    """
    Duration minutes, message count ve gb_used için plan bazında paylanma grafikleri
    """
    
    # Figure oluştur
    fig, axes = plt.subplots(3, 2, figsize=(16, 14))
    fig.suptitle('Plan Bazında Kullanım Dağılımları', fontsize=16, fontweight='bold', y=0.995)
    
    # Renkler
    colors = {'surf': '#3498db', 'ultimate': '#e74c3c'}
    
    # 1. Duration Minutes - Histogram
    ax1 = axes[0, 0]
    for plan in ['surf', 'ultimate']:
        data = final_df[final_df['plan'] == plan]['duration_minutes'].dropna()
        ax1.hist(data, bins=50, alpha=0.6, label=plan.capitalize(), color=colors[plan], edgecolor='black')
    ax1.set_xlabel('Duration Minutes', fontsize=11, fontweight='bold')
    ax1.set_ylabel('Frekans', fontsize=11, fontweight='bold')
    ax1.set_title('Çağrı Süresi Dağılımı (Histogram)', fontsize=12, fontweight='bold')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # 2. Duration Minutes - Boxplot
    ax2 = axes[0, 1]
    data_to_plot = [final_df[final_df['plan'] == 'surf']['duration_minutes'].dropna(),
                    final_df[final_df['plan'] == 'ultimate']['duration_minutes'].dropna()]
    bp1 = ax2.boxplot(data_to_plot, labels=['Surf', 'Ultimate'], patch_artist=True,
                      showmeans=True, meanline=True)
    for patch, color in zip(bp1['boxes'], [colors['surf'], colors['ultimate']]):
        patch.set_facecolor(color)
        patch.set_alpha(0.6)
    ax2.set_ylabel('Duration Minutes', fontsize=11, fontweight='bold')
    ax2.set_title('Çağrı Süresi Dağılımı (Boxplot)', fontsize=12, fontweight='bold')
    ax2.grid(True, alpha=0.3, axis='y')
    
    # 3. Message Count - Histogram
    ax3 = axes[1, 0]
    for plan in ['surf', 'ultimate']:
        data = final_df[final_df['plan'] == plan]['message_count'].dropna()
        ax3.hist(data, bins=50, alpha=0.6, label=plan.capitalize(), color=colors[plan], edgecolor='black')
    ax3.set_xlabel('Message Count', fontsize=11, fontweight='bold')
    ax3.set_ylabel('Frekans', fontsize=11, fontweight='bold')
    ax3.set_title('Mesaj Sayısı Dağılımı (Histogram)', fontsize=12, fontweight='bold')
    ax3.legend()
    ax3.grid(True, alpha=0.3)
    
    # 4. Message Count - Boxplot
    ax4 = axes[1, 1]
    data_to_plot = [final_df[final_df['plan'] == 'surf']['message_count'].dropna(),
                    final_df[final_df['plan'] == 'ultimate']['message_count'].dropna()]
    bp2 = ax4.boxplot(data_to_plot, labels=['Surf', 'Ultimate'], patch_artist=True,
                      showmeans=True, meanline=True)
    for patch, color in zip(bp2['boxes'], [colors['surf'], colors['ultimate']]):
        patch.set_facecolor(color)
        patch.set_alpha(0.6)
    ax4.set_ylabel('Message Count', fontsize=11, fontweight='bold')
    ax4.set_title('Mesaj Sayısı Dağılımı (Boxplot)', fontsize=12, fontweight='bold')
    ax4.grid(True, alpha=0.3, axis='y')
    
    # 5. GB Used - Histogram
    ax5 = axes[2, 0]
    for plan in ['surf', 'ultimate']:
        data = final_df[final_df['plan'] == plan]['gb_used'].dropna()
        ax5.hist(data, bins=50, alpha=0.6, label=plan.capitalize(), color=colors[plan], edgecolor='black')
    ax5.set_xlabel('GB Used', fontsize=11, fontweight='bold')
    ax5.set_ylabel('Frekans', fontsize=11, fontweight='bold')
    ax5.set_title('İnternet Kullanımı Dağılımı (Histogram)', fontsize=12, fontweight='bold')
    ax5.legend()
    ax5.grid(True, alpha=0.3)
    
    # 6. GB Used - Boxplot
    ax6 = axes[2, 1]
    data_to_plot = [final_df[final_df['plan'] == 'surf']['gb_used'].dropna(),
                    final_df[final_df['plan'] == 'ultimate']['gb_used'].dropna()]
    bp3 = ax6.boxplot(data_to_plot, labels=['Surf', 'Ultimate'], patch_artist=True,
                      showmeans=True, meanline=True)
    for patch, color in zip(bp3['boxes'], [colors['surf'], colors['ultimate']]):
        patch.set_facecolor(color)
        patch.set_alpha(0.6)
    ax6.set_ylabel('GB Used', fontsize=11, fontweight='bold')
    ax6.set_title('İnternet Kullanımı Dağılımı (Boxplot)', fontsize=12, fontweight='bold')
    ax6.grid(True, alpha=0.3, axis='y')
    
    plt.tight_layout()
    return fig


def plot_kde_distributions(final_df):
    """
    KDE (Kernel Density Estimation) ile daha düzgün dağılım grafikleri
    """
    
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    fig.suptitle('Plan Bazında Kullanım Dağılımları (KDE)', fontsize=16, fontweight='bold')
    
    colors = {'surf': '#3498db', 'ultimate': '#e74c3c'}
    
    # 1. Duration Minutes KDE
    ax1 = axes[0]
    for plan in ['surf', 'ultimate']:
        data = final_df[final_df['plan'] == plan]['duration_minutes'].dropna()
        data.plot(kind='kde', ax=ax1, label=plan.capitalize(), color=colors[plan], linewidth=2.5)
    ax1.set_xlabel('Duration Minutes', fontsize=11, fontweight='bold')
    ax1.set_ylabel('Yoğunluk', fontsize=11, fontweight='bold')
    ax1.set_title('Çağrı Süresi Dağılımı', fontsize=12, fontweight='bold')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # 2. Message Count KDE
    ax2 = axes[1]
    for plan in ['surf', 'ultimate']:
        data = final_df[final_df['plan'] == plan]['message_count'].dropna()
        data.plot(kind='kde', ax=ax2, label=plan.capitalize(), color=colors[plan], linewidth=2.5)
    ax2.set_xlabel('Message Count', fontsize=11, fontweight='bold')
    ax2.set_ylabel('Yoğunluk', fontsize=11, fontweight='bold')
    ax2.set_title('Mesaj Sayısı Dağılımı', fontsize=12, fontweight='bold')
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    
    # 3. GB Used KDE
    ax3 = axes[2]
    for plan in ['surf', 'ultimate']:
        data = final_df[final_df['plan'] == plan]['gb_used'].dropna()
        data.plot(kind='kde', ax=ax3, label=plan.capitalize(), color=colors[plan], linewidth=2.5)
    ax3.set_xlabel('GB Used', fontsize=11, fontweight='bold')
    ax3.set_ylabel('Yoğunluk', fontsize=11, fontweight='bold')
    ax3.set_title('İnternet Kullanımı Dağılımı', fontsize=12, fontweight='bold')
    ax3.legend()
    ax3.grid(True, alpha=0.3)
    
    plt.tight_layout()
    return fig





fig1 = plot_distribution_comparison(final_df)
plt.show()

fig2 = plot_kde_distributions(final_df)
plt.show()
```

```python
from scipy import stats

# Surf planı üçün   H0 normal paylanır
surf_cost = final_df[final_df['plan'] == 'surf']['total_cost']
shapiro_surf = stats.shapiro(surf_cost)

print("=" * 50)
print("SURF PLANI - Shapiro-Wilk Testi")
print("=" * 50)
print(f"Test statistikası: {shapiro_surf.statistic:.6f}")
print(f"P-value: {shapiro_surf.pvalue:.6e}")

if shapiro_surf.pvalue > 0.05:
    print("✅ Nəticə: Normal paylanma (p > 0.05) H0 not rejected")
else:
    print("❌ Nəticə: Normal paylanma DEYİL (p < 0.05) H0 rejected")

print("\n" + "=" * 50)
print("ULTIMATE PLANI - Shapiro-Wilk Testi")
print("=" * 50)

# Ultimate planı üçün  H0 normal paylanır
ultimate_cost = final_df[final_df['plan'] == 'ultimate']['total_cost']
shapiro_ultimate = stats.shapiro(ultimate_cost)

print(f"Test statistikası: {shapiro_ultimate.statistic:.6f}")
print(f"P-value: {shapiro_ultimate.pvalue:.6e}")

if shapiro_ultimate.pvalue > 0.05:
    print("✅ Nəticə: Normal paylanma (p > 0.05) H0 not rejected")
else:
    print("❌ Nəticə: Normal paylanma DEYİL (p < 0.05) H0 rejected")
```

```python
u_stat, p_value_mw = stats.mannwhitneyu(surf_cost, ultimate_cost, alternative='two-sided')

print("\n" + "=" * 60)
print("MANN-WHITNEY U TEST (Non-parametrik)")
print("=" * 60)
print(f"U-statistika: {u_stat:.4f}")
print(f"P-value: {p_value_mw:.6e}")
print(f"\nAlpha = 0.05 səviyyəsində:")

if p_value_mw < 0.05:
    print("✅ H₀ REDDEDİLİR: Qruplar ARASıNDA FƏRQ VAR")
    print("   → Surf və Ultimate gəlirləri fərqlidir")
else:
    print("❌ H₀ QƏBUL EDİLİR: Qruplar EYNİDİR")
    print("   → Surf və Ultimate gəlirləri statistik olaraq eynidir")
```

```python
from scipy import stats

# Mann-Whitney U Test
u_stat, p_value = stats.mannwhitneyu(
    ultimate_cost, 
    surf_cost, 
    alternative='less'  # H₁: Ultimate ≤ Surf
)

print("\n" + "=" * 70)
print("MANN-WHITNEY U TEST NƏTİCƏSİ")
print("=" * 70)
print(f"\nU-statistika: {u_stat:.4f}")
print(f"P-value: {p_value:.6e}")

# Qərar
print("\n" + "=" * 70)
print("QƏRAR")
print("=" * 70)

if p_value < 0.05:
    print(f"\n❌ P-value ({p_value:.6e}) < 0.05")
    print("\n❌ H₀ REDDEDİLİR, H₁ QƏBUL EDİLİR")
    print("\n📌 NƏTİCƏ:")
    print("   → Ultimate Surf-dən böyük DEYİL")
    print("   → Sizin iddia (Ultimate > Surf) SƏHVDIR")
else:
    print(f"\n✅ P-value ({p_value:.6e}) ≥ 0.05")
    print("\n✅ H₀ QƏBUL EDİLİR, H₁ REDDEDİLİR")
    print("\n📌 NƏTİCƏ:")
    print("   → H₀ (Ultimate > Surf) DOĞRUDUR")
    print(f"   → Ultimate orta hesabla {ultimate_cost.mean() - surf_cost.mean():.2f} USD daha çox gəlir gətirir")
    print("   → Bu fərq statistik olaraq ÖNƏMLİDİR")

print("\n" + "=" * 70)
print("🎯 YEKUN")
print("=" * 70)

if p_value >= 0.05:
    print("\n✅ SİZİN HİPOTEZİNİZ DOĞRUDUR!")
    print(f"   H₀: Ultimate > Surf ✓")
    print(f"   (p = {p_value:.6e} ≥ 0.05)")
else:
    print("\n❌ SİZİN HİPOTEZİNİZ SƏHVDIR!")
    print(f"   H₀: Ultimate > Surf ✗")
    print(f"   (p = {p_value:.6e} < 0.05)")

print("=" * 70)
```
